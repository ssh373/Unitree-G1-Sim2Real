```python
"""******************************************************************************************
* HumARConoid-Sim2Real
*
* Sim2Real Transfer to Optimize Humanoid Locomotion Strategy using Reinforcement Learning
*
*     https://github.com/S-CHOI-S/HumARConoid-Sim2Real.git
*
* Advanced Robot Control Lab. (ARC)
* 	  @ Korea Institute of Science and Technology
*
*	  https://sites.google.com/view/kist-arc
*
******************************************************************************************"""

"* Authors: Sol Choi (Jennifer) *"

import os
import numpy as np
import matplotlib.pyplot as plt

joint_names = [
    "LeftHipPitch", "LeftHipRoll", "LeftHipYaw", "LeftKnee", "LeftAnklePitch", "LeftAnkleRoll",
    "RightHipPitch", "RightHipRoll", "RightHipYaw", "RightKnee", "RightAnklePitch", "RightAnkleRoll",
    "WaistYaw", "WaistRoll", "WaistPitch",
    "LeftShoulderPitch", "LeftShoulderRoll", "LeftShoulderYaw", "LeftElbow", "LeftWristRoll", "LeftWristPitch", "LeftWristYaw",
    "RightShoulderPitch", "RightShoulderRoll", "RightShoulderYaw", "RightElbow", "RightWristRoll", "RightWristPitch", "RightWristYaw"
]
```

```python
'''  LEFT LEG  '''

data = np.load('log_data_simple_tutorial.npz')

cur_pos = data['cur_pos']
des_pos = data['des_pos']
time = data['time']

if isinstance(cur_pos, set):
    print("cur_pos is a set, converting to numpy array")
    cur_pos = np.array(list(cur_pos))

if isinstance(des_pos, set):
    print("des_pos is a set, converting to numpy array")
    des_pos = np.array(list(des_pos))

plt.figure(figsize=(10, 20))

print("elapsed time:", max(time), "sec")

for i in range(6):
    plt.subplot(6, 1, i+1)
    plt.plot(time, cur_pos[:, i], label=f"Current Position ({joint_names[i]})")
    plt.plot(time, des_pos[:, i], label=f"Desired Position ({joint_names[i]})", linestyle='-')
    plt.xlabel('Time (seconds)')
    plt.ylabel('Position')
    plt.title(f'{joint_names[i]}')
    plt.legend()
    plt.grid(True)

plt.tight_layout()
# plt.savefig("tracking_left_leg_tutorial.svg", format="svg")
plt.show()
```
