```python
'''  run func. dt  '''

data = np.load('log_data_simple_korea.npz')

dt = data['dt']

plt.figure(figsize=(10, 5))

print("mean:", "{:.10f}".format(np.mean(dt)), "sec,", f"max: {np.max(dt)} sec, min: {np.min(dt)} sec")
print(f"mean: {1/np.mean(dt)} Hz,", f"max: {1/np.max(dt)} Hz, min: {1/np.min(dt)} Hz")

print(f"{dt[:10]}")

plt.plot(range(len(dt)), dt, label="control dt", linestyle='None', marker='o', markersize=2)
plt.xlabel('Steps')
plt.ylabel('dt')
plt.ylim([0,0.03])

plt.title(f'func.[RUN] dt')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig("control_dt_tutorial.svg", format="svg")
plt.show()
```
